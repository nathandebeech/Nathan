```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
# 2.1.1.3
df = pd.read_csv('solar_system.csv')
df
```

```python
# 2.1.1.4
df1 = df.set_index("Attribute").T
df1.index.name = "Planet"
df1.reset_index(inplace=True)
df1.columns.name = None
df1
```

```python
# 2.1.1.5
# a.
print(df.shape)
print(df1.shape)
# b.
# 20 rows, 11 columns before. 10 rows, 21 columns afterward.
# c.
# The first line moves the attributes to the columns and the planets to the rows.
# The second line names the index numbers the name of each planet.
# The third line moves the planet names to its own column.
# The fourth line removes the name 'Atributes' from the index column.
# d.
# The first column of the new dataframe is the name of the planets versus the original dataframe that had the planet names as the name of each 
# column.
# e.
df1.columns
# f.
# 16 columns have units
# Columns with units: Mass, Diameter, Density, Gravity, Escape Velocity, Rotation Period, 
# Length of Day, Distance from Sun, Perihelion, Aphelion, Orbital Period, 
# Orbital Velocity, Orbital Inclination, Obliquity to Orbit, Mean Temperature, and 
# Surface Pressure.
# 4 columns do not have units
# Columns without units: Orbital Eccentricity, 
# Number of Moons, Ring System?, and Global Magnetic Field?
# g.
df1
```

```python
# 2.1.2.3
# a.
# Every value is a string
# b.
# By storing numbers as strings, we can not apply mathematical functions to the data.
```

```python
# 2.1.3 Convert Strings to Numbers
# 2.1.3.1
number_columns = ['Mass (10^24kg)', 'Diameter (km)', 'Density (kg/m^3)',
       'Gravity (m/s^2)', 'Escape Velocity (km/s)', 'Rotation Period (hours)',
       'Length of Day (hours)', 'Distance from Sun (10^6 km)',
       'Perihelion (10^6 km)', 'Aphelion (10^6 km)', 'Orbital Period (days)',
       'Orbital Velocity (km/s)', 'Orbital Inclination (deg)',
       'Orbital Eccentricity', 'Obliquity to Orbit (deg)',
       'Mean Temperature (C)', 'Surface Pressure (bars)', 'Number of Moons',]
for col in number_columns:
    df1[col] = pd.to_numeric(df1[col], errors="coerce")
```

```python
# 2.1.3.2
df1.dtypes
```

```python
# 2.1.3.3
# a.
# objects, floats, and integers.
# b.
# An object is anything, float string, etc. A float is a number with a decimal -- a 
# not whole number. An integer is a whole number
# c.
# The columns Planet, Ring System?, and Global Magnetic Field? are still objects becasue 
# we did not change their datatype due to the fact that they are words and no numbers.
```

```python
# 2.2 Apply Astropy Units
# 2.2.1
def attach_units(old_col):
    if old_col == 'Mass (10^24kg)':
        new_values = []
        new_col = 'Mass (kg)'
        for val in df1[old_col]:
            new_values.append(val * 1**24 * u.kg)
        df1[old_col] = new_values
        df1.rename(columns = {old_col: new_col}, inplace = True)
    elif old_col == 'Diameter (km)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.km)
        df1[old_col] = new_values
    elif old_col == 'Density (kg/m^3)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.kg / (u.m)**3)
        df1[old_col] = new_values
    elif old_col == 'Gravity (m/s^2)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.m / (u.s)**2)
        df1[old_col] = new_values
    elif old_col == 'Escape Velocity (km/s)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.km / u.s)
        df1[old_col] = new_values
    elif old_col == 'Rotation Period (hours)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.hour)
        df1[old_col] = new_values
    elif old_col == 'Length of Day (hours)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.hour)
        df1[old_col] = new_values
    elif old_col == 'Distance from Sun (10^6 km)':
        new_values = []
        new_col = 'Distance from Sun (km)'
        for val in df1[old_col]:
            new_values.append(val * 1**6 * u.km)     
        df1[old_col] = new_values
        df1.rename(columns = {old_col: new_col}, inplace = True)
        df1[old_col] = new_values
    elif old_col == 'Perihelion (10^6 km)':
        new_values = []
        new_col = 'Perihelion (km)'
        for val in df1[old_col]:
            new_values.append(val * 1**6 * u.km)     
        df1[old_col] = new_values
        df1.rename(columns = {old_col: new_col}, inplace = True)
        df1[old_col] = new_values
    elif old_col == 'Aphelion (10^6 km)':
        new_values = []
        new_col = 'Aphelion (km)'
        for val in df1[old_col]:
            new_values.append(val * 1**6 * u.km)     
        df1[old_col] = new_values
        df1.rename(columns = {old_col: new_col}, inplace = True)
        df1[old_col] = new_values
    elif old_col == 'Orbital Period (days)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.d)
        df1[old_col] = new_values
    elif old_col == 'Orbital Velocity (km/s)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.km / u.s)
        df1[old_col] = new_values
    elif old_col == 'Orbital Inclination (deg)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.deg)
        df1[old_col] = new_values
    elif old_col == 'Obliquity to Orbit (deg)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.deg)
        df1[old_col] = new_values
    elif old_col == 'Mean Temperature (C)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.deg_C)
        df1[old_col] = new_values
    elif old_col == 'Surface Pressure (bars)':
        new_values = []
        for val in df1[old_col]:
            new_values.append(val * u.bar)
        df1[old_col] = new_values



attach_units('Mass (10^24kg)')
attach_units('Diameter (km)')
attach_units('Density (kg/m^3)')
attach_units('Gravity (m/s^2)')
attach_units('Escape Velocity (km/s)')
attach_units('Rotation Period (hours)')
attach_units('Length of Day (hours)')
attach_units('Distance from Sun (10^6 km)')
attach_units('Perihelion (10^6 km)')
attach_units('Aphelion (10^6 km)')
attach_units('Orbital Period (days)')
attach_units('Orbital Velocity (km/s)')
attach_units('Orbital Inclination (deg)')
attach_units('Obliquity to Orbit (deg)')
attach_units('Mean Temperature (C)')
attach_units('Surface Pressure (bars)')
df1['Density (kg/m^3)']
```

```python
# 2.3 Data Analysis
# 2.3.1 Semi-Major Axis

perihelion = df1['Perihelion (km)']
aphelion = df1['Aphelion (km)']

semi_major_axis = (perihelion + aphelion) / 2
semi_major_axis

df1.insert(11, 'Semi Major Axis (km)', semi_major_axis)
```

```python
# 2.3.1 Continued
df1.columns
df1['Semi Major Axis (km)']
# Units!
```

```python
# 2.3.2 Convert Orbital Period to Years
df1.rename(columns = {'Orbital Period (days)': 'Orbital Period (years)'}, inplace = True)
converted_values = []
for value in df1['Orbital Period (years)']:
    converted_values.append(value.to(u.yr))
    
df1['Orbital Period (years)'] = converted_values

print(df1.columns)
print(df1['Orbital Period (years)'])
print(f"{df1['Orbital Period (years)'].iloc[1]:.4f}")
```

```python
# 2.3.3 Convert Distance Columns (km --> AU):
# 1 AU is equal to the average distance between the Sun and the Earth
print(const.au)
(const.au).to(u.km) # 1.4959787 * 10^8 km in 1 AU


distance_columns = ['Diameter (km)', 'Distance from Sun (km)', 
                    'Perihelion (km)', 'Aphelion (km)']
for column in distance_columns:
    if column == 'Diameter (km)':
        new_values = []
        new_col = 'Diameter (AU)'
        for val in df1[column]:
            new_values.append(val.to(u.au))
        df1[column] = new_values
        df1.rename(columns = {column: new_col}, inplace = True)
    elif column == 'Distance from Sun (km)':
        new_values = []
        new_col = 'Distance from Sun (AU)'
        for val in df1[column]:
            new_values.append(val.to(u.au))
        df1[column] = new_values
        df1.rename(columns = {column: new_col}, inplace = True)
    elif column == 'Perihelion (km)':
        new_values = []
        new_col = 'Perihelion (AU)'
        for val in df1[column]:
            new_values.append(val.to(u.au))
        df1[column] = new_values
        df1.rename(columns = {column: new_col}, inplace = True)
    elif column == 'Aphelion (km)':
        new_values = []
        new_col = 'Aphelion (AU)'
        for val in df1[column]:
            new_values.append(val.to(u.au))
        df1[column] = new_values
        df1.rename(columns = {column: new_col}, inplace = True)
```
