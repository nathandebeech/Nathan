```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",skiprows=1)
data
```

```python
# 3.2.1 
# 2642 rows
data.shape
```

```python
# 3.2.3
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
# 3.2.4
print(max(magnitude))
print(min(magnitude))
print(np.argmax(magnitude))  # The 264th star
print(np.argmin(magnitude))  # The 1988th star
```

```python
# 3.2.5 
print(max(temperature))
print(min(temperature))
print(np.argmax(temperature))  # The 2613th star
print(np.argmin(temperature))  # The 1923th star
```

```python
# 3.2.6
# The 1988th is the brightest and the 264th is the dimmest
```

```python
# 3.3.1
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
# 3.3.2
x = np.linspace(0, 2 *np.pi, 100)
y = np.sin(x)
plt.figure(figsize=(8, 6))
plt.plot()
plt.show()
```

```python
# 3.3.3
plt.scatter(log_temperature, log_magnitude)
plt.show()
```

```python
# 3.3.4.
plt.scatter(log_temperature, log_magnitude)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.show()
```

```python
# 3.3.5.
plt.scatter(log_temperature, log_magnitude)
plt.scatter(log_temperature[1988], log_magnitude[1988], color='black', marker='*', s=250)
plt.scatter(log_temperature[264], log_magnitude[264], color='purple', marker='*', s=250)
plt.scatter(log_temperature[1923], log_magnitude[1923], color='orange', marker='*', s=250)
plt.scatter(log_temperature[2613], log_magnitude[2613], color='green', marker='*', s=250)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.show()
```

```python
# 3.3.6.
blue_dots = plt.scatter(log_temperature, log_magnitude)
black_star = plt.scatter(log_temperature[1988], log_magnitude[1988], color='black', marker='*', s=250)
purple_star = plt.scatter(log_temperature[264], log_magnitude[264], color='purple', marker='*', s=250)
orange_star = plt.scatter(log_temperature[1923], log_magnitude[1923], color='orange', marker='*', s=250)
green_star = plt.scatter(log_temperature[2613], log_magnitude[2613], color='green', marker='*', s=250)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.legend([blue_dots, black_star, purple_star, orange_star, green_star], ["Stars!", "Brightest Star", "Dimmest Star", "Coldest Star", "Hottest Star"])

plt.show()
```

```python
# 3.3.7 and 8 and 9
dots = plt.scatter(log_temperature, log_magnitude, c= log_temperature, cmap='plasma_r', vmax= 9)
plt.colorbar(label= "Coldest - Hotest")
black_star = plt.scatter(log_temperature[1988], log_magnitude[1988], color='black', marker='*', s=250)
purple_star = plt.scatter(log_temperature[264], log_magnitude[264], color='purple', marker='*', s=250)
orange_star = plt.scatter(log_temperature[1923], log_magnitude[1923], color='orange', marker='*', s=250)
green_star = plt.scatter(log_temperature[2613], log_magnitude[2613], color='green', marker='*', s=250)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.legend([dots, black_star, purple_star, orange_star, green_star], ["Stars", "Brightest Star", "Dimmest Star", "Coldest Star", "Hottest Star"])
plt.title('Hertzsprung_russel Diagram')
plt.xlabel('Temperature (log K)')
plt.ylabel('Magnitude')

plt.show()
```
