```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2.1 

def make_sine_wave(x, A, w):
    y = A * np.sin(w * x)
    return y
```

```python
# 2.2 
x = np.linspace(0, 2 *np.pi, 1000)
```

```python
# 2.3
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
# 2.4
plt.subplots(1, 3, figsize=(25, 5))
plt.show()
```

```python
# 2.5
for A, w in zip(amplitude, frequency):
    print(A, w)
```

```python
# 2.6
fig, axes = plt.subplots(1, 3, figsize=(25, 5))
for A, w in zip(amplitude, frequency):
    axes[0].plot(make_sine_wave(x, A, 1))
    axes[1].plot(make_sine_wave(x, 1, w))
    axes[2].plot(make_sine_wave(x, A, w))
plt.show()
```

```python
# 2.7 and 8
fig, axes = plt.subplots(1, 3, figsize=(25, 5))
for A, w in zip(amplitude, frequency):
    axes[0].plot(make_sine_wave(x, A, 1))
    axes[1].plot(make_sine_wave(x, 1, w))
    axes[2].plot(make_sine_wave(x, A, w))
    axes[0].set_title("Sine Wave: Varying A", fontsize=28)
    axes[1].set_title("Sine Wave: Varying w", fontsize=28)
    axes[2].set_title("Sine Wave: Varying A and w", fontsize=28)
    axes[0].set_xlabel("Radians", fontsize=20)
    axes[0].set_ylabel("Amplitdue", fontsize=20)
    axes[1].set_xlabel("Radians", fontsize=20)
    axes[1].set_ylabel("Amplitude", fontsize=20)
    axes[2].set_xlabel("Radians", fontsize=20)
    axes[2].set_ylabel("Amplitude", fontsize=20)
plt.show()
```
