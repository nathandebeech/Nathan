```python
# 4 Prime Star Systems
```

```python
magnitudes = np.array([    
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])
i = 2
for row in magnitudes:
    for number in row:
        is_prime = True
        for i in range(2, int(number ** 0.5) + 1):
              if number % i == 0:
                  is_prime = False
    if is_prime:
        print(row)
```

```python
# 5.2
```

```python
board = np.zeros((8,8))
for i in range(board.shape[0]):
    if i % 2 != 0:       
        for j in range(board.shape[1]):
            if (i+j) % 2 == 0:
                board[i,j] = 1
print(board)
```

```python
# 5.3
```

```python
board = np.zeros((8,8))
for i in range(board.shape[0]):
    for j in range(board.shape[1]):
        if (i+j) % 2 == 0:
            board[i,j] = 1
print(board)
```

```python
# 5.4
```

```python
for row in board:
    print(row[::-1])
```

```python
# 6 The Expanding Unverse
```

```python
universe = np.array(['galaxy', 'clusters'])

def expansion(array, a):
    spaces = a * ' '
    for word in array:
        expanded_word = spaces.join(word)
        print(expanded_word)
expansion(universe, 4)
```
