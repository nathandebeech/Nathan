```python
# 4 Prime Star Systems
```

```python
magnitudes = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])
prim_list = []
i = 2
for row in magnitudes:
    for number in row:
          if number == 2:
              prime_list.append(number)
          elif number > 2:
               while i < number:
                   if number % i != 0:
                      prime_list.append(number)
                        i +=1
print(magnitudes)
```
