```python
import numpy as np
```

```python
# 3.1
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)
```

```python
print(arr * 4)
```

```python
print(np.mean(arr))
```

```python
print(np.median(arr))
```

```python
print(np.std(arr))
```

```python
print(np.argmin(arr))
print(np.argmax(arr))
```

```python
print(arr[6])
print(arr[7])
```

```python
print(np.min(arr))
print(np.max(arr))
# They match above!
```

```python
greater_than_twenty = arr > 20
new_array = arr[greater_than_twenty]
print(new_array)
```
