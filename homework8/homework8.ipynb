```python
# 3. Seaborn
# 3.4
df1 = sns.load_dataset("exercise")
df1
```

```python
# 3.5
df1 = df1.drop(df1.columns[0], axis=1)
```

```python
# 3.6
df1.shape
# 90 rows and 6 columns
df1.head()
# The id column is identifies a specific individual doing these exercises
# the diet column lists what type of diet the individual is on during this exercise
# the pulse column is the measure of the individual's heart rate
# the time column is the durration of the exercise
# the kind column is the type of exercise the individual completes
```

```python
# 3.6
df1.tail()
```

```python
# 3.7 and .8
x = df1.pulse
y = df1.time
fig, axes = plt.subplots(2,2, figsize = (13,10))
axes[0, 0].scatter(x,y, c= 'orange', label = "Exercise")
axes[0, 0].legend()
axes[0, 0].set_xlabel('Pulse (bpm)')
axes[0, 0].set_ylabel('Time (min)')
axes[0, 0].set_title('Pulse vs Duration')
axes[0, 1].scatter(y,x, c= 'brown')
axes[0, 1].set_xlabel('Time (min)')
axes[0, 1].set_ylabel('Pulse (bpm)')
axes[0, 1].set_title('Pulse Over Time')
sns.histplot(x, ax=axes[1,0], color= 'skyblue')
axes[1,0].set_title('Pulse of Exercises')
axes[1,0].set_xlabel('Pulse (bpm)')
axes[1,0].set_ylabel('Number of "Pulses"')
sns.boxplot(y, ax=axes[1,1])
axes[1,1].set_title('Time Data')
axes[1,1].set_xlabel('Number of "Times"')
axes[1,1].set_ylabel('Time (min)')
plt.show()
```
