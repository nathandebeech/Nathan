```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
# 2.1 Collect the Data
df = pd.read_csv("../../../python_decal_fa25/course_assignments/homework9/noisy_curvefit_data.csv")
df
```

```python
# 2.1.4
print(df.shape)
# 100 rows and 6 columns
print(df.head())
print(df.tail())
print(df.columns)
# Columns: 'x', 'y_noise_1', 'y_noise_2', 'y_noise_3', 'y_noise_4', 'y_noise_5'
```

```python
# 2.1.5
x = df['x']
y1 = df['y_noise_1']
plt.figure(figsize=(6,6))
plt.scatter(x, y1, c='gray', label= 'Noise1')
plt.xlabel('x label')
plt.ylabel('Noise level 1')
plt.legend()
plt.title('Noise graph 1')
plt.show()
```

```python
# 2.2.1
# model is y = A * e ** (-k * x)
def model(x, A, k):
    return A * np.exp(-k * x)
```

```python
# 2.2.2
p0 = [1,1]
popt, pcov = curve_fit(model, x, y1, p0=p0)
```

```python
# 2.2.3
A_fit, k_fit = popt
```

```python
# 2.2.4
print(f"Fit: y = {A_fit:.2f} * e^(-{k_fit:.2f} *x)")
```

```python
# 2.2.5
residuals = y1 - model(x, A_fit, k_fit)
chi2 = np.sum(residuals**2)
dof = len(x) - len(popt)
chi2_reduced = f"{chi2 / dof:.2f}"      # Rounded reduced chi squared
```

```python
# 2.2.6
plt.figure()
plt.scatter(x, y1, c= "gray", label = 'Noise1')
plt.plot(x, model(x, 4.71, 0.92), c = "red", label = 'Fitted model')
plt.plot(x, model(x, 5, 0.8), c = "black", linestyle = "--", label = 'True model')
plt.xlabel("X Label")
plt.ylabel("Noise level 1")
plt.legend()
plt.title("Noise level 1 with Reduced chi_squared")
plt.show()
```

```python
# 2.3 Compare Different Noise Levels
# 2.3.1 
def fit_model(x, y):
    popt, pcov = curve_fit(model, x, y, p0=p0)
    A_fit, k_fit = popt
    f"Fit: y = {A_fit:.2f} * e^(-{k_fit:.2f} *x)"
    residuals = y - model(x, A_fit, k_fit)
    chi2 = np.sum(residuals**2)
    dof = len(x) - len(popt)
    chi2_reduced = f"{chi2 / dof:.2f}"
    return chi2_reduced and popt
fit_model(x, y1)
```

```python
# 2.3.2-3
y2 = df['y_noise_2']
y3 = df['y_noise_3']
y4 = df['y_noise_4']
y5 = df['y_noise_5']
noise_levels = [y2, y3, y4, y5]

fit_model2 = fit_model(x, y2)
fit_model3 = fit_model(x, y3)
fit_model4 = fit_model(x, y4)
fit_model5 = fit_model(x, y5)
fit_models = [fit_model2, fit_model3, fit_model4, fit_model5]

titles = ["Noise Level 2 with Reduced Chi_squared", "Noise Level 3 with Reduced Chi_squared", "Noise Level 4 with Reduced Chi_squared", "Noise Level 5 with Reduced Chi_squared"]
xlabels = ["x values", "x values", "x values", "x values"]
ylabels = ["Noise level 2", "Noise level 3", "Noise level 4", "Noise level 5"]

fig, axes = plt.subplots(2, 2, figsize=(12,10))
for ax, y_noise, title, xlabel, ylabel, in zip(axes.ravel(), noise_levels, titles, xlabels, ylabels):
    A = fit_model(x, y_noise)[0]
    k = fit_model(x, y_noise)[1]
    ax.scatter(x, y_noise, c = 'gray')
    ax.plot(x, model(x, A, k), c = 'red')
    ax.plot(x, model(x, 5, 0.8), c = 'black', linestyle = '--', label = 'True Model')  
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel) 

plt.show()
```

```python
# 2.3.4
The dotted black and red line begin to match more closely
```
